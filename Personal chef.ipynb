```python
from dotenv import load_dotenv

load_dotenv()
```

```python
#Step-1: Prepare model

from langchain_ollama import ChatOllama
from langchain_core.messages import HumanMessage, SystemMessage

# Initialize the ChatOllama model instance
model = ChatOllama(
    model="gpt-oss:20b",
    temperature=0.7,
    validate_model_on_init=True,
    base_url="http://localhost:11434"
)
```

```python
#Create tools for agent

from langchain.tools import tool
from tavily import TavilyClient


@tool
def search_web(query: str) -> str:
    ''' This will help in searching the web for information'''
    tavily_client = TavilyClient()

    return tavily_client.search(query)
    
```

```python
#Step-2: Creating agent with tools and memory

from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver  


system_prompt='''
You are a personal chef. You need to help a bachelor in cooking. The bachelor has PCOS issue.The bachelor lady name is Hari K Priya and you need to assiste her in making simple 
and healthy dishes. She has limited dishes. She will give you the list of items she has in her kitchen. You need to provide recipies with available items. 
She has PCOS and prefers to eat healthy and avoid sugar spikes. Provide the recipie along with ingredients and stepwise procedure

Use search web tool to search for recipies in web 
'''
chef=create_agent(model=model,
                  system_prompt=system_prompt,
                  checkpointer=InMemorySaver(),
                 tools=[search_web])
```

```python
#Step-3: Invoke agent

from langchain.messages import HumanMessage

question = HumanMessage(content="I have eggs, milk, bread, mushroom, spinach, bitter guard, peanut butter, rice, bottle gourd. Help me prepare breakfast")
config = {"configurable": {"thread_id": "1"}}

response = chef.invoke(
    {"messages": [question]},
    config,  
)
```
